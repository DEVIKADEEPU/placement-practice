```python
rat = int(input("no of rats: "))
unit = int(input("food consumed by each rat: "))
n = int(input("no of houses: "))
l = list(map(int, input("food in each house: ").split()))

if l is None:
    print(-1)
else:
    totalfood = rat * unit
    food = 0
    houses = 0
    for i in range(n):
        food = food + l[i]
        houses = houses + 1

        if food >= totalfood:
            break

    if food < totalfood:
        print(0)
    else:
        print(houses)
```
